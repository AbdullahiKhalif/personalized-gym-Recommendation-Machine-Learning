# Personalized Gym Recommmendation System

** The "Personal Gym Recommendation" project aims to build a system that helps gym-goers get a recommendation system that takes into account the specific fitness preferences for beginners **

The dataset contains the following variables:

* **'1. ID':**  Unique identifier for each user..
* **'2. Sex':** Indicates the gender of the individual (Male/Female).
* **'3. Age':** Age of the individual in years.
* **'4. Height':** Height of the individual in meters.
* **'5. Weight':** Weight of the individual in kilograms.
* **'6. Hypertension':** Indicates whether the individual has hypertension (Yes/No).
* **'7. Diabetes':** Indicates whether the individual has diabetes (Yes/No).
* **'8. BMI':** Body Mass Index of the individual, calculated as weight in kilograms divided by the square of height in meters.
* **'9. Level':** : Categorizes the individual's weight status based on BMI (e.g., Underweight, Normal, Overweight, Obese).
* **'10. Lower Limmit':** (possibly meant to be "Lower Limit"): Lower limit of the standard weight range for the individual, based on height.
* **'11. Upper Limmit':** (possibly meant to be "Upper Limit"): Upper limit of the standard weight range for the individual, based on height.
* **'12. Fat Percentage':** The percentage of the individual's weight that is fat.
* **'13. Fitness Goal':** The individual's fitness goal (e.g., Weight Gain, Weight Loss).
* **'14. Fitness Type:':** : Type of fitness the program is targeting (e.g., Muscular Fitness, Cardiovascular Fitness).
* **'15. Exercises':** Recommended exercises for the individual to meet their fitness goal.
* **'16. Equipment Required':** HThe equipment needed to perform the recommended exercises.
* **'17. Diet':** Recommended diet or foods for the individual to follow, aimed at supporting their fitness goal.
* **'18. Conclusion Recommendation':** A general conclusion or recommendation given to the individual based on their profile and fitness goals.

## Checking outliers

```python
# Define a function to detect outliers based on the IQR method
def detect_outliers_iqr(data, column):
    Q1 = data[column].quantile(0.25)
    Q3 = data[column].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    return df[(df[column] < lower_bound) | (df[column] > upper_bound)].shape[0]

# Columns to check for outliers
columns_to_check = ['Age', 'Height', 'Weight', 'BMI', 'Lower Limmit', 'Upper Limmit', 'Fat Percentage']

# Calculate the number of outliers in each column
outliers_count = {column: detect_outliers_iqr(df, column) for column in columns_to_check}
outliers_count
```

----------------------

## Drop All Outliers from dataset

```python
# Define a function to remove outliers based on the IQR method for a given column
def remove_outliers_iqr(data, column):
    Q1 = data[column].quantile(0.25)
    Q3 = data[column].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    return data[(df[column] >= lower_bound) & (df[column] <= upper_bound)]

# Apply the outlier removal for each column individually
df = df.copy()
for column in columns_to_check:
    df = remove_outliers_iqr(df, column)

# Check the new shape of the dataset to see how many records remain
df.shape
```

## Conclusion

### Project Description
The Personal Gym Recommendation System (PGRS) is a digital platform designed to provide tailored fitness and health advice to gym-goers, especially those new to the gym environment. By analyzing user-provided data such as age, height, weight, and medical conditions like hypertension and diabetes, PGRS crafts personalized fitness and dietary plans aligned with individual health profiles and goals.

### Problem Statement
The current generation is increasingly concerned about their health and fitness. However, new gym-goers often do not know their fitness goals clearly, what exercises they should do, and what equipment to use, leading to inefficient workout routines and delayed progress. For example, someone aiming to lose weight might engage in exercises better suited for gaining weight, risking serious injury. Many individuals struggle to achieve their fitness goals due to the limitations of existing personalized fitness recommendations. The main problem is that everyone wants the coach to track their progress or provide recommendations, which is a significant burden for gym coaches.

### Research Objectives
- **General Objective**: The main goal of our study is to create a system that assists people who go to the gym by providing them with a recommendation system.
- **Specific Objectives**:
  - To develop a gym recommendation system that takes into account the specific fitness preferences of both beginners and intermediates.
  - To recommend tailored fitness programs, including specific exercises and equipment required based on individuals' unique health profiles and fitness goals.
  - To design a user-friendly interface where users can enter their information, such as personal profiles or preferences.

### Our Research Contribution
Our research contribution involves adding three new columns to the dataset: Lower Limit, Upper Limit, and Body Fat Percentage. These additions provide a deeper analysis of each individual's fitness and health metrics.
1. **Lower Limit**: This column represents the lower boundary of a healthy weight range for each person, calculated using the formula:  
   **`Lower Limit = 18.5 * height(m)^2`**  
   Here, **18.5** is the lower threshold of the normal BMI range, multiplied by the individual's height in meters squared. This helps in identifying if a person is underweight.
2. **Upper Limit**: Similar to the Lower Limit, the Upper Limit indicates the upper boundary of a healthy weight range, calculated as:  
   **`Upper Limit = 24.9 * height(m)^2`**  
   The value **24.9** is the upper threshold of the normal BMI range, again multiplied by the person's height in meters squared. This measure helps identify if a person is approaching an overweight range.
3. **Body Fat Percentage**: This column provides an estimate of the body fat percentage, a key indicator of fitness. For men, it is calculated using the formula:  
   **`(1.20 x BMI) + (0.23 x Age) – 16.2`**  
   For women, the formula is slightly different:  
   **`(1.20 x BMI) + (0.23 x Age) – 5.4`**  
   These formulas account for differences in body composition between men and women and provide a more personalized assessment of body fat levels.

### Our Research Gap
Our research uncovered some gaps that need addressing to enhance the effectiveness and applicability of our gym recommendation system:
1. **Limited Scope of Study**: Our current system primarily caters to individuals who are new to the gym. This means it provides general recommendations suitable for beginners, focusing on basic fitness routines and exercises. However, it does not account for the needs of more experienced gym-goers or those with specific fitness goals like advanced strength training, bodybuilding, or endurance sports. This limitation can make the recommendations less relevant or beneficial for users with different levels of experience or specialized training needs. Expanding the system to include tailored advice for a broader range of gym-goers, from beginners to advanced athletes, would make it more versatile and useful.
2. **Limited Scope of Health**: Currently, our system includes considerations for two specific health conditions: hypertension and diabetes. While these are important and common conditions, this limited scope overlooks other health issues that could significantly impact a person's fitness regimen. Conditions such as heart disease, respiratory issues, joint problems, or previous injuries require different types of exercise advice and modifications. By only focusing on hypertension and diabetes, we may not be providing safe or effective recommendations for users with other health conditions. To improve, the system should incorporate a wider range of medical conditions, allowing for more comprehensive and personalized fitness plans.

3. **Some Accuracy Models are Low**: While some machine learning models in our system, like Decision Trees and Random Forests, showed high accuracy in making recommendations, others did not perform as well. For instance, models like K-Nearest Neighbors (KNN) and Support Vector Machines (SVM) had lower accuracy rates, indicating they might not be as effective in predicting the best gym routines for users. These lower accuracy rates suggest that the models may not be capturing all relevant factors or that they are not the best fit for our dataset. Improving the accuracy of these models is crucial to ensure that all users receive reliable and effective fitness recommendations. This improvement could involve optimizing the current models, exploring alternative algorithms, or enhancing the dataset with more detailed and diverse information.
